```python
# Analisis statistika deskriptif performa pembelajaran pemrograman dasar
# Kebutuhan: pandas, numpy, matplotlib (dijalankan di Google Colab)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 0. Mengunggah data dari komputer (Google Colab)
from google.colab import files
uploaded = files.upload()  # pilih classification_data.csv dari komputer
```

```python
# 1. Membaca dan memeriksa data
df = pd.read_csv("classification_data.csv", sep=";")
print(df.shape)
print(df.dtypes)
print(df.isna().sum())
print(df.describe().T)

utama = ["lab_1", "delivery_time_lab_1", "attempts_lab_1"]
semua = utama + ["lab_2", "lab_3"]
label = {"lab_1": "Nilai Lab 1",
         "delivery_time_lab_1": "Waktu Pengumpulan Lab 1",
         "attempts_lab_1": "Jumlah Percobaan Lab 1"}

# 2. Distribusi frekuensi dan persentase grade
frek = df["grade"].value_counts().sort_index()
print(pd.DataFrame({"frekuensi": frek, "persen": (frek / len(df) * 100).round(2)}))

# 3. Pemusatan, penyebaran, bentuk sebaran, dan batas pencilan
hasil = {}
for c in semua:
    s = df[c]
    q1, q2, q3 = s.quantile([0.25, 0.50, 0.75])
    iqr = q3 - q1
    bb, ba = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    hasil[c] = {"mean": s.mean(), "median": q2, "modus": s.mode().iloc[0],
                "min": s.min(), "max": s.max(), "range": s.max() - s.min(),
                "Q1": q1, "Q3": q3, "IQR": iqr, "varians": s.var(), "sd": s.std(),
                "skewness": s.skew(), "excess_kurtosis": s.kurt(),
                "batas_bawah": bb, "batas_atas": ba,
                "di_bawah": int((s < bb).sum()), "di_atas": int((s > ba).sum())}
print(pd.DataFrame(hasil).round(3))

# 4. Ringkasan per kategori grade
print(df.groupby("grade")[semua].agg(["count", "mean", "median", "std"]).round(2))

# 5. Visualisasi
WARNA = "#2A6F97"
WARNA_GRADE = {0: "#D9822B", 1: "#2A9D8F", 2: "#264653"}

def simpan(fig, nama):
    fig.savefig(nama, dpi=200, bbox_inches="tight", facecolor="white")
    plt.close(fig)

# 6a. Diagram batang grade
fig, ax = plt.subplots(figsize=(6, 3.8))
ax.bar([f"Grade {g}" for g in frek.index], frek.values,
       color=[WARNA_GRADE[g] for g in frek.index])
ax.set_ylabel("Jumlah data")
simpan(fig, "gambar_4_1_grade.png")

# 6b. Histogram
fig, axs = plt.subplots(1, 3, figsize=(12, 3.8))
batas = {"lab_1": np.arange(0, 5.5, 0.5),
         "delivery_time_lab_1": np.arange(0, 23, 1.0),
         "attempts_lab_1": np.arange(0, 42, 2)}
for ax, c in zip(axs, utama):
    ax.hist(df[c], bins=batas[c], color=WARNA, edgecolor="white")
    ax.axvline(df[c].mean(), color="#C0392B", ls="--", label="Mean")
    ax.axvline(df[c].median(), color="#222222", label="Median")
    ax.set_title(label[c]); ax.set_ylabel("Frekuensi"); ax.legend(frameon=False)
fig.tight_layout(); simpan(fig, "gambar_4_2_histogram.png")

# 6c. Boxplot keseluruhan
fig, axs = plt.subplots(1, 3, figsize=(10, 4))
for ax, c in zip(axs, utama):
    ax.boxplot(df[c], widths=0.45)
    ax.set_title(label[c]); ax.set_xticks([])
fig.tight_layout(); simpan(fig, "gambar_4_3_boxplot.png")

# 6d. Boxplot per kategori grade
fig, axs = plt.subplots(1, 3, figsize=(12, 4.2))
for ax, c in zip(axs, utama):
    ax.boxplot([df[df["grade"] == g][c] for g in (0, 1, 2)],
               tick_labels=["Grade 0", "Grade 1", "Grade 2"])
    ax.set_title(label[c])
fig.tight_layout(); simpan(fig, "gambar_4_4_boxplot_grade.png")
```
